```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = "../data/west_bengal_climate_final.nc"

ds = xr.open_dataset(DATA_PATH)
print(ds)
```

```python
print(ds.dims)
```

```python
print(ds.coords)
```

```python
print(ds.data_vars)
```

```python
for var in ["rainfall", "tmax", "tmin"]:
    print("\n", var)
    print(ds[var])
```

```python
for var in ["rainfall", "tmax", "tmin"]:
    print(
        var,
        "min =", float(ds[var].min()),
        "max =", float(ds[var].max()),
        "mean =", float(ds[var].mean())
    )
```

```python
print(ds.time.values[0])
print(ds.time.values[-1])
print(len(ds.time))
```
